# Stroke Volume Quantification

Loads histologically-derived stroke injury volumes from `stroke_summary.csv` and plots **total injury volume (mm³)** across four experimental groups: CS Young Female, CS Young Male, WMS Young Female, WMS Young Male.

Group names in the CSV are heterogeneous (e.g., 'yf_cls', 'cls young female') and are normalized to a canonical form by `standardize_group()` before plotting.

**Output:** a bar plot (68% CI) with individual data points overlaid, saved as a PDF, plus printed summary statistics (mean, SEM, count) per group.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Read the CSV file
df = pd.read_csv('./stroke_summary.csv')

# Create the plot
plt.figure(figsize=(10, 6),dpi=300)

def standardize_group(group):
    group = group.lower()
    if any(x in group for x in ['cls young female', 'yf_cls']):
        return 'CS Young Female'
    elif 'cls young male' in group or 'ym_cls' in group:
        return 'CS Young Male'
    elif 'wms young female' in group or 'yf_wms' in group:
        return 'WMS Young Female'
    elif 'wms young male' in group or 'ym_wms' in group:
        return 'WMS Young Male'
    return group

# Standardize group names
df['group'] = df['group'].apply(standardize_group)
sns.set_style('white')

# Create bar plot with individual points
group_order = ['CS Young Female', 'CS Young Male', 'WMS Young Female', 'WMS Young Male']
colors = ['lightpink', 'lightblue', 'red', 'blue']

# Create bar plot with individual points
ax = sns.barplot(x='group', y='total injury volume (mm^3)', data=df,
                order=group_order,
                palette=colors,
                alpha=0.5,
                ci=68)  # Using 68% CI (standard error)

# Add individual data points
sns.stripplot(x='group', y='total injury volume (mm^3)', data=df,
             order=group_order,
             color='black', alpha=0.4, jitter=0.2, size=7)


# Rotate x-axis labels for better readability
plt.xticks(rotation=45, ha='right')

# Adjust layout to prevent label cutting
plt.tight_layout()

# Show the plot
plt.savefig('/content/drive/My Drive/MoSeq_Figures/Supplementary/Supplementary1/stroke_size_quantification_barplot.pdf')
plt.show()

# Print summary statistics
print("\nSummary Statistics:")
summary = df.groupby('group')['total injury volume (mm^3)'].agg(['mean', 'sem', 'count'])
print(summary)
```
